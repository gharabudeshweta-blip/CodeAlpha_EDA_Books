```python
# 1. Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#2. Load Dataset
df = pd.read_csv("books_data.csv")
df.head()
```

```python
# 4. Data Cleaning
df['Price'] = df['Price'].str.replace('Â|£', '', regex=True).astype(float)
```

```python
print(df)
```

```python
# 5. Exploratory Data Analysis
# 6. Data Visualization (Matplotlib Charts)
df['Rating'].value_counts().plot(kind='bar')

plt.title("Book Ratings Distribution")
plt.xlabel("Rating")
plt.ylabel("Count")

plt.show()
```

```python
plt.hist(df['Price'], bins=10)

plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Number of Books")

plt.show()
```

```python
df.groupby('Rating')['Price'].mean().plot(kind='bar')

plt.title("Average Price by Rating")
plt.xlabel("Rating")
plt.ylabel("Average Price")

plt.show()
```

```python
# 📌 Insights from Analysis

- The dataset contains book information including title, price, and rating.
- Most books belong to the medium rating category, showing average customer feedback.
- The distribution of ratings helps understand overall customer satisfaction.
- Book prices vary across different ratings.
- Higher-rated books do not always have the highest prices, which shows that price depends on other factors as well.


# ✅ Conclusion

In this Exploratory Data Analysis project, I analyzed a book dataset using Python, Pandas, and Matplotlib.

The analysis helped understand patterns in book ratings and prices. Data visualization made it easier to identify trends and compare different categories.

This project improved my understanding of:
- Data cleaning
- Data exploration
- Statistical analysis
- Data visualization techniques

The insights obtained can help in making better decisions based on book pricing and customer ratings.
```
