```python
import pandas as pd
```

```python

import matplotlib.pyplot as plt

df = pd.read_csv("books_data.csv")
df.head()
```

```python
df.info()
```

```python
df.tail()
```

# Data Cleaning

```python
df['Price'] = df['Price'].astype(str)
```

```python
df['Price'] = df['Price'].str.replace('Â', '')
df['Price'] = df['Price'].str.replace('£', '')
```

```python
df['Price'] = df['Price'].astype(float)
df
```

```python
rating_map = {
    "One":1, "Two":2, "Three":3, "Four":4, "Five":5
}
df['Rating'] = df['Rating'].map(rating_map)
```

```python
df
```

```python
rating_counts = df['Rating'].value_counts().sort_index()

plt.figure()
plt.bar(rating_counts.index, rating_counts.values)
plt.xlabel("Rating")
plt.ylabel("Number of Books")
plt.title("Distribution of Book Ratings")
plt.show()
```

```python
plt.figure()
plt.hist(df['Price'], bins=10)
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Price Distribution of Books")
plt.show()
```

```python
plt.figure()
plt.scatter(df['Rating'], df['Price'])
plt.xlabel("Rating")
plt.ylabel("Price")
plt.title("Price vs Rating")
plt.show()
```

```python
plt.figure()
plt.hist(df['Price'], bins=10)
plt.title("Price Distribution of Books")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.grid()
plt.tight_layout()
plt.show()
```

## 📊 Insights

- Most books are rated between 3 and 4 stars  
- Prices are mostly in the range of £10–£40  
- There is no strong relationship between price and rating  

## ✅ Conclusion

The dataset shows that moderately priced books dominate the market, and higher prices do not necessarily mean better ratings.
